```python
_2022_prod = _2022.groupby('product_id')
```

```python
_2022_prod_sale = {}
```

```python
for pid in _2022_prod.product_id.unique():
    print(f"Total sales of {pid[0]}: {_2022_prod.get_group(pid[0]).quantity.sum()}")
    _2022_prod_sale[pid[0]] = _2022_prod.get_group(pid[0]).quantity.sum()
```

```python
_2022_prod_sale = pd.Series(_2022_prod_sale)
```

```python
print(f"Most Saling product in 2022 by quantity: {_2022_prod_sale.index[_2022_prod_sale.argmax()]}")
```

```python
for year in orders.year.unique():
    year_data = grouped_orders.get_group(year)
    prod_group = year_data.groupby('product_id')
    prod_sales = {}
    for pid in prod_group.product_id.unique():
        prod_sales[pid[0]] = prod_group.get_group(pid[0]).quantity.sum()
    prod_sales = pd.Series(prod_sales)
    print(f"Most Saling product in {year} by quantity: {prod_sales.index[prod_sales.argmax()]}")
```

```python
for year, year_data in grouped_orders:
    best = year_data.groupby('product_id')["quantity"].sum().idxmax()
    print(f"Most Saling product in {year} by quantity: {best}")
```
