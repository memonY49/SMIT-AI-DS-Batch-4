```python
for year in orders.year.unique():
    year_data = grouped_orders.get_group(year)
    prod_group = year_data.groupby('product_id')
    prod_sales = {}
    for pid in prod_group.product_id.unique():
        prod_sales[pid[0]] = prod_group.get_group(pid[0]).quantity.sum()
    prod_sales = pd.Series(prod_sales)
    print(f"Most Saling product in {year} by quantity: {prod_sales.index[prod_sales.argmax()]}")
```

```python
new_data = pd.merge(orders,products, on="product_id")
```

```python
new_data["total_price"] = new_data.price * new_data.quantity
```

```python
new_data
```

```python
grouped_data = new_data.groupby('year')
```

```python
for year, df in grouped_data:
    print(f"Sales in {year}: {df.total_price.sum()}")
```

```python
for year, df in grouped_data:
    print("-"*10,year,"-"*10)
    g_category = df.groupby('category')
    total_sales_c = g_category['total_price'].sum()
    print(total_sales_c)
    print(f"Most Salling category is: {total_sales_c.idxmax()}")
```

```python
new_data.pivot_table(
    values='total_price',
    index='year',
    columns='category',
    aggfunc='sum',
    fill_value=0
)
```
