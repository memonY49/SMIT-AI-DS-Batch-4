# Logistic Regression

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris, load_breast_cancer
```

```python
data = load_iris()
data
```

```python
df = pd.DataFrame(data['data'], columns=data['feature_names'])
```

```python
df['species'] = data['target']
```

```python
df = df[df.species != 2]
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
LogisticR = LogisticRegression()
```

```python
LogisticR.fit(x_train,y_train)
```

```python
y_test_pred = LogisticR.predict(x_test)
```

```python
y_test
```

```python
y_test_pred
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report
```

```python
import seaborn as sns
```

```python
sns.heatmap(confusion_matrix(y_test, y_test_pred), annot=True)
```

```python
accuracy_score(y_test,y_test_pred)
```

```python
precision_score(y_test,y_test_pred)
```

```python
print(classification_report(y_test,y_test_pred))
```

```python
y_train_pred = LogisticR.predict(x_train)
```

```python
print(classification_report(y_train,y_train_pred))
```

```python
cancer = load_breast_cancer()
```

```python
cancer_df = pd.DataFrame(cancer["data"], columns=cancer["feature_names"])
```

```python
cancer_df['target'] = cancer['target']
```

```python
cancer_df
```

```python
from sklearn.linear_model import RidgeClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
param = {"alpha":[1,2,3,0.5,4],
        "tol":[0.001,0.0001,0.00001,0.000001]}
RC_grid = GridSearchCV(
    estimator=RidgeClassifier(),
    param_grid=param,
    scoring='accuracy',
    cv=5
)
```

```python
RC_grid_pipline = Pipeline(
    [("Standerd Scaler",StandardScaler()),
     ("Grid Search Cv",RC_grid)]
)
```

```python
x_train,x_test,y_train,y_test = train_test_split(cancer_df.iloc[:,:-1],cancer_df.iloc[:,-1], test_size=0.2,random_state=42)
```

```python
RC_grid_pipline.fit(x_train,y_train)
```

```python
y_train_pred = RC_grid_pipline.predict(x_train)
y_test_pred = RC_grid_pipline.predict(x_test)
```

```python
print(f"traning accuracy: {accuracy_score(y_train,y_train_pred)}")
print(f"testing accuracy: {accuracy_score(y_test,y_test_pred)}")
```
