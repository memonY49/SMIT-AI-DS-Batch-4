```python
import pandas as pd
import seaborn as sns
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = load_diabetes()
data
```

```python
dia = pd.DataFrame(data['data'], columns=data['feature_names'])
dia
```

```python
dia['target'] = data['target']
```

```python
dia
x = dia.iloc[:,:-1]
y = dia.iloc[:,-1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.20, random_state=42)
```

```python
LR = LinearRegression(tol=0.00001)
```

```python
LR.fit(x_train,y_train)
```

```python
y_test_pred = LR.predict(x_test)
```

```python
y_test_pred
```

```python
y_test
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mean_absolute_error(y_test,y_test_pred)
```

```python
house_data = pd.read_csv('housing_price_dataset.csv')
```

```python
house_data
```

```python
sns.heatmap(house_data.corr(), annot=True)
```

```python
house_data.dropna(inplace=True)
house_data.drop("Neighborhood", axis=1, inplace=True)
```

```python
x = house_data.iloc[:,:-1]
y = house_data.iloc[:, -1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.3, random_state=42)
```

```python
LR = LinearRegression()
```

```python
LR.fit(x_train,y_train)
```

```python
y_test_pred = LR.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,root_mean_squared_error,r2_score
```

```python
print("MAE: ",mean_absolute_error(y_test,y_test_pred))
print("RMSE: ",root_mean_squared_error(y_test,y_test_pred))
print("R2_Score: ",r2_score(y_test,y_test_pred))
```
