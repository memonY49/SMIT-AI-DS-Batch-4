````python
studentdata = [["Math","Scince","English","Urdu","Sindhi"],
               [50,60,70,80,40],
               [30,70,80,40,60],
               [20,40,60,80,90],
               [30,60,80,50,40],
               [40,40,30,60,80]]
subject = studentdata.pop(0)
````

````python
studentdata
````

````python
# Find Total Obtain Marks
# Find Percentage for every student by (obtain/total)100
# Find Grade on (per>=90 -> A, per>=80 and <90 -> B, per>=70 and <80 -> C, per>=60 and <70 -> D, per<60 -> Fail)
# Assign Ranking on percentage.
````

````python
std_data = np.array(studentdata)
````

````python
obtain = std_data.sum(axis=1)
````

````python
per = obtain / 500 * 100
````

````python
# grade = np.where(per>=90,'A','B')
# grade = np.where(per>=80,'B',grade)
# grade = np.where(per>=70,'C',grade)
# grade = np.where(per>=60,'D',grade)
# grade = np.where(per<60,'Fail',grade)
````

````python
# grade
````

````python
per
````

````python
# per.sort()
# per[::-1]
````

````python
rank = per.argsort()[::-1]
rank
````

````python
for r,std in enumerate(rank):
    grade = ''
    if per[std]>= 90:
        grade = 'A'
    elif per[std]>=80 and per[std] <90:
        grade = 'B'
    elif per[std]>=70 and per[std] <80:
        grade = 'C'
    elif per[std]>=60 and per[std] <70:
        grade = 'D'
    else:
        grade = 'Fail'
    print(f"Student no: {std+1}")
    print(f"Rank no:{r+1}")
    print(f"Obtain: {obtain[std]}")
    print(f"Percentage: {per[std]}")
    print(f"Grade: {grade}")
    # print(f"Grade: {grade[std]}")
    print("*"*10)
````
