```python
students = pd.read_csv("students_five_classes.csv")
```

```python
students["obtain"] = students.iloc[:,2:].sum(axis = 1)
```

```python
students['total'] = 100
```

```python
students['per'] = students.obtain/students.total *100
```

```python
def grade(marks):
    if marks>80:
        return 'A'
    elif marks>70:
        return 'B'
    elif marks>50:
        return 'C'
    else:
        return 'F'
```

```python
students["grade"] = students.per.apply(grade)
```

```python
students['class'] = students.roll_no.str.split('/').str[1]
```

```python
grade_count = students.grade.value_counts()
```

```python
plt.bar(x = grade_count.index, height= grade_count.values)
plt.bar(x,y)
# plt.bar_label(bar)
```

```python
students
```

```python
gr_data = students.groupby('class')
```

```python
# sub = plt.subplot(3,2,1)
for cls, c in zip(students['class'].unique(),['green','blue','red','pink','brown']):
    data = gr_data.get_group(cls)
    counts = data.grade.value_counts()
    plt.bar(x = counts.index, height = counts.values, label = cls, color = c)
    plt.title(cls)
    plt.xlabel("Grade")
    plt.ylabel("Counts")
    plt.legend()
   
```

```python
std_pivot = students.pivot_table(
    index='grade',
    columns = 'class',
    aggfunc='size'
)
std_pivot
```

```python
plt.figure(figsize=(10,12))
plot = std_pivot.plot(kind = 'bar',subplots = True,layout = (2,3),figsize = (10,10))
# for i in plot.containers:
#     plt.bar_label(i)
```
