```python
area_sqyr = [60,120,150,180,200,250,300]
price = [35,60,90,100,140,160,200]
```

```python
plt.plot(area_sqyr,price,c = "#ff5291",label = "Price pre", marker = 'x', linestyle='dotted')
plt.title("Price over area in sqyr",pad=10)
plt.ylabel("Price in Lac",labelpad=10)
plt.xlabel("Area in sqyr")
plt.text(100,125,"Hello",color='#fcb0ff')
plt.legend()
plt.savefig("Linechart_price_sqyr.png")
plt.show()
```

```python
y1 =[]
y2 =[]
x = range(-100,100,10)
for i in x: 
    y1.append(i**2) 
for i in x: 
    y2.append(-i**2) 
```

```python
y2
```

```python
plt.figure(figsize=(4,4))
plt.plot(x,y1, marker = 'o')
plt.plot(x,y2, marker = 'x')
plt.axvline(x=0, c = 'g')
plt.axhline(y = 0, c = 'g')
plt.ylim(-8000,8000)
plt.xlim(-100,100)
plt.grid(True,linewidth=3,axis='y')
```
