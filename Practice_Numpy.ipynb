## An online store records daily sales (in dollars) for 3 products over 7 days.
# Tasks
1. Find total weekly sales for each product.
2. Find the day with the highest overall sales.
3. Calculate average daily sales across all products.
4. Identify products whose weekly sales exceed $1000.
5. Apply a 10% discount to all sales values greater than 200.

````python
import numpy as np
sales = np.array([
    [120, 150, 180, 200, 170, 190, 210],  # Product A
    [80, 90, 100, 110, 95, 105, 115],     # Product B
    [200, 220, 210, 230, 240, 250, 260]   # Product C
])
````

````python
weekly_sales_prod = sales.sum(axis = 1)
daily_sales = sales.sum(axis = 0)
highest_sales_day = daily_sales.argmax()+1
prod_average_sales = sales.mean(axis=1)
prod_sales_1000 = np.where(weekly_sales_prod > 1000)
discounted = np.where(sales > 200, sales - (sales * 0.1), sales)
````

````python
print(f"Weekly sales per prod: {weekly_sales_prod}")
print(f"Highest sales on day {highest_sales_day}")
print(f"Average daily sales per prod: {prod_average_sales}")
print(f"Prod with weekly sales exceeded by 1000$ : {prod_sales_1000[0]+1}")
print(discounted)
````

## An investor starts with an initial investment of $5000. The yearly profit/loss percentages for the next 5 years are:
# Tasks
1. Convert the profit/loss percentages into growth factors.
    1. Example: 20% → 1.20, -20% → 0.80
2. Calculate the investment value at the end of each year.
3. Find the total profit or loss amount after 5 years.
4. Calculate the overall profit/loss percentage relative to the initial investment.
5. Determine:
    1. Which year had the highest growth?
    2. Which year had the biggest loss?
6. Create an array showing the profit/loss amount earned in each year.
   
- Expected Result
- Year-end values:
- Year 1: 6000
- Year 2: 7800
- Year 3: 8580
- Year 4: 6864
- Year 5: 6177.60
- Final investment value: 6177.60
- Total profit: 1177.60
- Overall return: 23.55%

````python
import numpy as n
initial_investment = 5000
profit_percent = np.array([20, 30, 10, -20, -10])
````

````python
g_factor = (profit_percent / 100) +1
g_factor
````

````python
profit_factor = g_factor.cumprod()
profit_factor
````

````python
yearly_investment = profit_factor * initial_investment
yearly_investment
````

````python
profit_year5 = yearly_investment[-1] - initial_investment
profit_year5
````

````python
total_profit_per = profit_year5 / initial_investment * 100
````

````python
print("Year End Values:")
for index, value in enumerate(yearly_investment):
    print(f"Year {index+1}: {value}")
print(f"final investment value: {yearly_investment[-1]}")
print(f"final profit: {profit_year5}")
print(f"Overall returns: {total_profit_per}")
````
