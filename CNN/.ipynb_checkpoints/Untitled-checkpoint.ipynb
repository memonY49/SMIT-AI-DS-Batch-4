```python
cv2.imwrite("elongray.png", gray)
```

```python
frontal_face = cv2.CascadeClassifier(cv2.data.haarcascades+"haarcascade_frontalface_default.xml")
```

```python
faces = frontal_face.detectMultiScale(
    gray,
    scaleFactor = 1.1,
    minNeighbors = 5,
    minSize = (30,30)
)
```

```python
faces
```

```python

for x,y,w,h in faces:
    cv2.rectangle(img,(x,y),(x+w, y+h),
             (0,255,0),2)
```

```python
cv2.imshow("my_image_window",img)
# cv2.imshow("grayImage", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)
frontal_face = cv2.CascadeClassifier(cv2.data.haarcascades+"haarcascade_frontalface_default.xml")
# eye_face = 
frame_no = 0
while True:
    ret, frame = cap.read()
    if not ret:
        print("cam not detected...!!!")
        break
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = frontal_face.detectMultiScale(
        gray_frame,
        scaleFactor = 1.1,
        minNeighbors = 1,
        minSize = (50,50)
    )
    display = frame.copy()
    for i,(x,y,w,h) in enumerate(faces):
        face_crop = frame[y:y+h,x:x+w]
        # frame_no += i
        # cv2.imwrite(f"faces/img{frame_no}.png", face_crop)
        cv2.rectangle(display,(x,y),(x+w, y+h),(0,255,0),2)
        
    combineframe = np.hstack((frame,display))
    combineframe = cv2.resize(combineframe, (1080,512))
    cv2.imshow("Live image", combineframe)
    
    if cv2.waitKey(1) == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
    
```
